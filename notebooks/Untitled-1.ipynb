```python
import json
from pathlib import Path

SCHEMA_FOLDER = Path("/mnt/e/Sanjana_Code/ShopSphere/data/table_schema")
OUTPUT_FILE = Path("/mnt/e/Sanjana_Code/ShopSphere/agents/sql/table_schema.json")
```

```python

database_schema = {
    "database": "shopsphere",
    "dialect": "PostgreSQL",
    "version": "1.0",
    "tables": [],
    "relationships": []
}

# Read every json file
for file in sorted(SCHEMA_FOLDER.glob("*.json")):
    with open(file, "r", encoding="utf-8") as f:
        columns = json.load(f)

    table_schema = {
        "table_name": file.stem.replace("_schema",""),          # customers.json -> customers
        "description": "",
        "columns": []
    }

    for col in columns:
        table_schema["columns"].append({
            "name": col["Column Name"],
            "type": col["Data Type"],
            "nullable": col["Nullable"]
        })

    database_schema["tables"].append(table_schema)

# Save combined schema
with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
    json.dump(database_schema, f, indent=4)

print(f"Saved schema with {len(database_schema['tables'])} tables.")
```
